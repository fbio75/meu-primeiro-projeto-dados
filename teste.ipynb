```python
import pandas as pd

# Criando dados de exemplo
dados_epi = {
    'Colaborador': ['João Silva', 'Maria Souza', 'Carlos Oliveira', 'Ana Costa'],
    'EPI': ['Bota de Segurança', 'Luva Química', 'Capacete', 'Máscara Respiratória'],
    'Data_Entrega': ['2026-04-01', '2026-04-10', '2026-04-15', '2026-04-20'],
    'Validade_Dias': [180, 90, 365, 30]
}

# Criando a tabela (DataFrame)
df_epi = pd.DataFrame(dados_epi)

# Salvando como um arquivo de Excel (XLSX)
df_epi.to_excel('controle_epis.xlsx', index=False)

print("Arquivo 'controle_epis.xlsx' criado com sucesso na sua pasta!")
```

```python
import pandas as pd
from datetime import timedelta

# 1. Ler o arquivo que criamos
df = pd.read_excel('controle_epis.xlsx')

# 2. Transformar a coluna de data para o formato que o Python entende
df['Data_Entrega'] = pd.to_datetime(df['Data_Entrega'])

# 3. Calcular a Data de Vencimento (Data de Entrega + Dias de Validade)
df['Data_Vencimento'] = df.apply(lambda x: x['Data_Entrega'] + timedelta(days=x['Validade_Dias']), axis=1)

# 4. Salvar o arquivo atualizado
df.to_excel('controle_epis.xlsx', index=False)

print("Cálculo de vencimento concluído e arquivo atualizado!")
```
